## <span style="color: green">Task 1: Business objectives and hypotheses</span>

### Use Case 1: Mietpreisvorhersage Kanton Zürich
1. **Problem:** Mieten sind schwer einzuschätzen. Falsche Preise führen zu Leerstand oder entgangenem Ertrag.
2. **Ziel:** Modell, das die Monatsmiete aus Wohnungsmerkmalen vorhersagt.
3. **Forschungsfrage:** Welche Merkmale beeinflussen die Miete am stärksten, und wie genau ist die Vorhersage?
4. **Hypothese:** H1: Wohnfläche, Zimmerzahl und Nähe zur Stadt Zürich erhöhen die Miete signifikant. H0: Kein Einfluss.
5. **Mehrwert:** Datenbasierte Preise für Vermietende, faire Preise erkennbar für Mietende.

### Use Case 2: Angebot und Nachfrage bei Supermärkten
1. **Problem:** Supermärkte sind ungleich verteilt. Manche Gemeinden sind unterversorgt, andere übersättigt.
2. **Ziel:** Supermarktangebot und Einwohnerzahl je Gemeinde vergleichen.
3. **Forschungsfrage:** Wie unterscheidet sich die Supermarktdichte (pro 1000 Einwohner) zwischen Gemeinden, und welche sind unterversorgt?
4. **Hypothese:** H1: Ländliche Gemeinden haben weniger Supermärkte pro Einwohner als städtische. H0: Kein Unterschied.
5. **Mehrwert:** Bessere Standortwahl für Detailhändler, Versorgungslücken für Gemeinden erkennbar.


## <span style="color: green">Task 2: Suitable data</span>

### Use Case 1: Mietpreisvorhersage Kanton Zürich

**immoscout24.ch (pro Inserat)**
- Zielvariable: Monatsmiete
- Wohnfläche (m²), Zimmerzahl, Stockwerk
- PLZ, Gemeinde, Strasse
- Baujahr, Objekttyp, Ausstattung (Balkon, Lift, Parkplatz)

**BFS Regionalporträts 2021 (pro Gemeinde)**
- Einwohnerzahl, Bevölkerungsdichte, Fläche
- Altersstruktur, Haushaltsgrösse, Ausländeranteil
- Leerwohnungsziffer, Neubautätigkeit
- Steuerfuss, Beschäftigte

**Verknüpfung:** Inserat über PLZ oder Gemeindename mit der BFS Gemeindenummer verbinden.

### Use Case 2: Supermärkte in Schweizer Gemeinden

**OpenStreetMap (`shop=supermarket`, Overpass API)**
- Koordinaten (lat, lon), ID
- Name, Marke (Migros, Coop, Denner, Aldi, Lidl)
- Adresse (Stadt, PLZ, Strasse)
- Öffnungszeiten, Rollstuhlzugang

**BFS Regionalporträts 2021 (pro Gemeinde)**
- Nachfrage: Einwohnerzahl, Haushalte, Altersstruktur, Bevölkerungsdichte
- Struktur: Beschäftigte, Raumtyp (städtisch oder ländlich), Fläche

**Verknüpfung:** Supermärkte über Koordinaten oder PLZ den Gemeinden zuordnen, pro Gemeinde zählen und mit BFS Daten zusammenführen. Daraus entsteht die Supermarktdichte (pro 1000 Einwohner).

**Hinweise**
- OSM Daten sind nicht überall vollständig (z. B. fehlt die Verkaufsfläche).
- Beim Scraping von immoscout24 die Nutzungsbedingungen beachten.
